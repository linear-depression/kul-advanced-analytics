```python
import matplotlib.pyplot as plt
from collections import Counter
import seaborn as sns
df = df.dropna(subset=['themes'])

# Extract all themes from the lists
all_themes = [theme for theme_list in df['themes'] for theme in theme_list]
    
    # Count frequencies
theme_counts = Counter(all_themes)
    
    # Convert to a DataFrame for easy plotting
counts_df = pd.DataFrame.from_dict(theme_counts, orient='index', columns=['count'])
counts_df = counts_df.sort_values('count', ascending=False)

print(f"Total unique themes found: {len(counts_df)}")
print("\nTop 20 Themes:")
print(counts_df.head(20))

    # Plot the Long-Tail distribution
plt.figure(figsize=(15, 6))
sns.barplot(x=counts_df.index[:30], y=counts_df['count'][:30], palette='viridis')
plt.title("Top 30 Lego Themes by Frequency (The Long Tail)")
plt.xticks(rotation=45, ha='right')
plt.ylabel("Number of Minifigures")
plt.xlabel("Theme name")
plt.tight_layout()
plt.savefig("../outputs/plots/theme_frequency.png", bbox_inches='tight')
plt.show()
```
